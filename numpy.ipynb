```python
import numpy as np

# From Python Lists
arr = np.array([1, 2, 3])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:")
print(arr)

print("\n2D Array:")
print(arr_2d)

# Special Arrays
print("\nZeros Array:")
print(np.zeros((3, 3)))

print("\nOnes Array:")
print(np.ones((2, 4)))

print("\nIdentity Matrix:")
print(np.eye(3))

print("\nArange:")
print(np.arange(0, 10, 2))

print("\nLinspace:")
print(np.linspace(0, 1, 5))

print("\nRandom Uniform Numbers:")
print(np.random.rand(3, 2))

print("\nRandom Normal Numbers:")
print(np.random.randn(3, 2))
```

```python
import numpy as np

# Arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a + b =", a + b)
print("a * b =", a * b)
print("sin(a) =", np.sin(a))

# Matrix Multiplication
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("\nnp.dot(A, B):")
print(np.dot(A, B))

print("\nA @ B:")
print(A @ B)

# Aggregation Functions
arr = np.array([1, 2, 3, 4])

print("\nSum:", arr.sum())
print("Mean:", arr.mean())
print("Max:", arr.max())
print("Min:", arr.min())
print("Standard Deviation:", arr.std())
```
